Proyecto Práctico: Desarrollo del Módulo Central de Autenticación y Registro

Como parte de la primera fase para la construcción del proyecto final de la asignatura, desarrollaremos la estructura base de un sistema de gestión de usuarios mediante una interfaz interactiva por consola. Este componente representa el punto de acceso principal de cualquier aplicación y requiere procesar, almacenar y proteger la información que ingresan los clientes de forma limpia y organizada.

El objetivo principal es construir una aplicación que se mantenga en ejecución continua a través de un menú de opciones, permitiendo registrar nuevos perfiles y consultar la información almacenada en tiempo real. Debido a que el sistema interactúa directamente con personas, la aplicación no debe detenerse bruscamente si alguien comete un error al digitar sus datos; por el contrario, debe reaccionar de manera segura cuando detecte texto en campos que requieren estrictamente números, como la edad o el código PIN de acceso, solicitando la información nuevamente hasta que sea correcta.

Para lograr un diseño estructurado, el programa debe combinar diferentes formas de organizar la información. Los roles dentro de la plataforma (ADMIN, ESTUDIANTE, DOCENTE) formarán un conjunto de opciones fijas que no cambiarán durante la ejecución. Por su parte, la información individual de cada usuario se organizará asociando cada dato con su respectiva etiqueta (nombre, edad, PIN y rol), para luego agrupar a todos los usuarios registrados dentro de un gran listado general que permitirá recorrer y consultar las cuentas creadas hasta que el operador decida cerrar la sesión del programa.

Opciones del menu:

2. Listar usuarios registrado

```python
def ValidarEntero(mensaje):
 while True:
  try:
    valor = int(input(mensaje))
    return valor
  except ValueError:
    print("Debe ingresar un numero entero, no aceptamos Caracteres!!!")
```

```python
def logeo(usuarios):
  #Diferentes Escenarios al momento de ingresar a un siste:  Diego
  #1 Uruario No existe
  #2 Contraseña Invalada
  #3 preguntar si desea registrarse.
  nombre = input("Ingrese usuario: ")
  clave = input("Ingrese clave: ")
  usuario_encontrado ={}
  existe_usuario = False
  for usuario in usuarios:
    if usuario['Nombre'] == nombre:
      existe_usuario = True
      usuario_encontrado = usuario
      break
  if existe_usuario == False:
    print("El usuario no existe.")
    opcion = input("¿Desea registrarse? (si/no): ")
    if opcion == "si":
      RegistrarUsuario(usuarios)
    return usuario_encontrado,2
  elif usuario_encontrado['Clave'] != clave:
    print("Contraseña invalada.")
    return usuario_encontrado,3
  else:
    print(f"Bienvenido al sistema {nombre}.")
    return usuario_encontrado,1

```

```python
def RegistrarUsuario(usuarios):
  print("Registrando usuario")
  nombre = input("Ingrese nombre de usuario: ")
  Edad= ValidarEntero("Ingrese edad: ")
  PIN = ValidarEntero("Ingrese PIN numerico: ")
  clave = input("Ingrese clave")
  rol = input("Ingrese rol: (Invitado-Admin-Vendedor) ").upper()
  usuarios.append({'Nombre':nombre,'Edad':Edad,'Pin':PIN,'Clave':clave,'Rol':rol})
  print(f"-> ¡Usuario '{nombre}' registrado con éxito!")
  return usuarios
```

```python
def ListarUsuarios(usuarios):
  print("Listando usuarios registrados")
  for usuario in usuarios:
    print(f"Nombre: {usuario['Nombre']}, Edad: {usuario['Edad']}, PIN: {usuario['Pin']}, Clave: {usuario['Clave']}, Rol: {usuario['Rol']}")
```
